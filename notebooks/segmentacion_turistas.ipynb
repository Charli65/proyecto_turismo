```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler

# Cargar datos
df = pd.read_csv('../data/turistas.csv')
df.head()
```

```python
# --- Preprocesamiento ---
X = df[['visitantes', 'gasto_promedio']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
# --- Segmentación con KMeans ---
kmeans = KMeans(n_clusters=3, random_state=42)
df['kmeans_segmento'] = kmeans.fit_predict(X_scaled)

sns.scatterplot(x='visitantes', y='gasto_promedio', hue='kmeans_segmento', data=df, palette='Set1')
plt.title('Segmentación de turistas con KMeans')
plt.show()
```

```python
# --- Segmentación con DBSCAN ---
dbscan = DBSCAN(eps=1.2, min_samples=2)
df['dbscan_segmento'] = dbscan.fit_predict(X_scaled)

sns.scatterplot(x='visitantes', y='gasto_promedio', hue='dbscan_segmento', data=df, palette='tab10')
plt.title('Segmentación de turistas con DBSCAN')
plt.show()
```
