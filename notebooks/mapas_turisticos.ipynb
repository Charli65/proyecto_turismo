```python
import pandas as pd
import folium

# Cargar datos
df = pd.read_csv('../data/turistas.csv')
df.head()
```

```python
# Diccionario con coordenadas aproximadas de ciudades principales
coords = {
    'Asunción': [-25.2637, -57.5759],
    'Encarnación': [-27.3306, -55.8667],
    'Ciudad del Este': [-25.5167, -54.6167]
}

# Crear mapa base
m = folium.Map(location=[-25.5, -57.0], zoom_start=6)

# Agregar marcadores por ciudad y tipo de turismo
for idx, row in df.iterrows():
    ciudad = row['ciudad']
    tipo = row['tipo_turismo']
    visitantes = row['visitantes']
    gasto = row['gasto_promedio']
    
    if ciudad in coords:
        folium.CircleMarker(
            location=coords[ciudad],
            radius=10 + visitantes/5000,
            popup=f"{ciudad} - {tipo}<br>Visitantes: {visitantes}<br>Gasto promedio: {gasto} USD",
            color='blue' if tipo=='Cultural' else 'green' if tipo=='Compras' else 'red',
            fill=True,
            fill_opacity=0.6
        ).add_to(m)

m
```

```python
# Guardar el mapa en outputs
m.save('../outputs/mapa_turismo.html')
print('Mapa guardado en outputs/mapa_turismo.html')
```
